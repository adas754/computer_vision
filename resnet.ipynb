```python
# Import ResNet50 model
from tensorflow.keras.applications.resnet50 import ResNet50

# Import image processing utilities
from tensorflow.keras.preprocessing import image

# Import preprocessing and prediction decoding functions
from tensorflow.keras.applications.resnet50 import preprocess_input, decode_predictions

# Import NumPy
import numpy as np
```

```python
model = ResNet50(weights='imagenet')
```

```python
# Set the path of the image
img_path = '/content/download.jpg'

# Load the image and resize it to 224x224 pixels
img = image.load_img(img_path, target_size=(224, 224))

# Convert the image into a NumPy array
x = image.img_to_array(img)


# Add an extra dimension for the batch because ResNet50 expects a batch of images
x = np.expand_dims(x, axis=0)

# Preprocess the image for ResNet50
x = preprocess_input(x)

# Make predictions
preds = model.predict(x)

# Display the top 3 predictions
print('Predicted:', decode_predictions(preds, top=3)[0])
```
