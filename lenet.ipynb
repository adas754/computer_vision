```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, AveragePooling2D, Flatten, Dense

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Make images 32x32 and add channel
x_train = tf.image.resize(x_train[..., None], (32, 32))
x_test = tf.image.resize(x_test[..., None], (32, 32))

# Normalize pixels
x_train = x_train / 255.0
x_test = x_test / 255.0

# Create LeNet model
model = Sequential()

model.add(Conv2D(6, (5,5), activation='tanh', input_shape=(32,32,1)))
model.add(AveragePooling2D((2,2), strides=2))

model.add(Conv2D(16, (5,5), activation='tanh'))
model.add(AveragePooling2D((2,2), strides=2))

model.add(Flatten())

model.add(Dense(120, activation='tanh'))
model.add(Dense(84, activation='tanh'))

model.add(Dense(10, activation='softmax'))

# Show model
model.summary()

# Compile
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train
model.fit(x_train, y_train, epochs=5, batch_size=128)

# Test
loss, accuracy = model.evaluate(x_test, y_test)

print("Test Accuracy:", accuracy)
```
